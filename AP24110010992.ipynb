### #10. Broadcasting

**Predection** :

Shape: (3, 3)

Values:

[[11 21 31]

 [12 22 32]

 [13 23 33]]

```python
#AP24110010992

import numpy as np                                                 # Import NumPy
result = np.array([[1], [2], [3]]) + np.array([10, 20, 30])        # Add the two arrays
print(result)                                                      # Display the result
```

**Explanation** :I create two arrays with different shapes. NumPy uses broadcasting to match them automatically. Each number in the first column is added to all three numbers in the second array, giving us a 3 by 3 matrix

**Reflection** : Broadcasting is a feature in NumPy that allows arrays with different shapes to work together in an operation. NumPy automatically expands the smaller array to match the larger array. For example, if we add [1, 2, 3] to 10, NumPy adds 10 to every element and gives [11, 12, 13].

### #11.Filtering Data (Boolean Indexing)

```python
import numpy as np                              # Import NumPy library
marks = np.array([45, 67, 89, 32, 90, 55])      # Create an array containing marks
print(marks[marks < 40])                        # Extract marks below 40
print((marks > 80).sum())                       # Count how many marks are above 80
new_marks = np.where(marks < 40, 0, marks)      # Replace marks below 40 with 0
print(new_marks)                                # Display the updated marks
```

**Explanation** : I create an array containing the students' marks. Then I use Boolean indexing to find the marks below 40. Next, I use sum() to count how many students scored above 80. Finally, I use np.where() to replace all failing marks with zero

### #14: Image Analysis Using NumPy

```python
from google.colab import files
uploaded = files.upload()
from PIL import Image                          # Import Image from PIL
import matplotlib.pyplot as plt                # Import Matplotlib
import numpy as np                             # Import NumPy
img_array = np.array(Image.open("naruto.jpeg"))  # Load image and convert it to an array
plt.imshow(img_array)                          # Display the image
plt.axis("off")                                # Hide the axes
plt.show()                                     # Show the image
```

**Explanation** :I import the Image library to open the image and Matplotlib to display it. Then I open the image and convert it into a NumPy array. I use imshow() to display the image, axis('off') to hide the axes, and show() to show the final image

**Image Cropping** :

```python
import matplotlib.pyplot as plt              # Import Matplotlib
height, width = img_array.shape[:2]          # Get the height and width of the image
start_y = (height - 200) // 2                # Calculate the starting row for the center
start_x = (width - 200) // 2                 # Calculate the starting column for the center
crop = img_array[start_y:start_y + 200,      # Crop 200 rows
                 start_x:start_x + 200]      # Crop 200 columns
plt.imshow(crop)                             # Display the cropped image
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
```

**Explanation** : I find the height and width of the image. Then I calculate the starting position of the center. Using slicing, I take a 200 by 200 region from the center and store it in crop. Finally, I display the cropped image and hide the axes

**Image Flipping** :

```python
import matplotlib.pyplot as plt              # Import Matplotlib
vertical_flip = img_array[::-1, :]           # Flip the image vertically
horizontal_flip = img_array[:, ::-1]         # Flip the image horizontally
plt.imshow(vertical_flip)                    # Display the vertical flip
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
plt.imshow(horizontal_flip)                  # Display the horizontal flip
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
```

**Explanation** : For the vertical flip, I reverse the rows of the image. For the horizontal flip, I reverse the columns. I use slicing instead of a built-in flip function, and then I display both images

**RGB Channel Analysis** :

```python
import matplotlib.pyplot as plt             # Import Matplotlib
red = img_array[:, :, 0]                    # Extract the Red channel
green = img_array[:, :, 1]                  # Extract the Green channel
blue = img_array[:, :, 2]                   # Extract the Blue channel
plt.imshow(red, cmap="gray")                # Display the Red channel
plt.title("Red Channel")                    # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
plt.imshow(green, cmap="gray")              # Display the Green channel
plt.title("Green Channel")                  # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
plt.imshow(blue, cmap="gray")               # Display the Blue channel
plt.title("Blue Channel")                   # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
```

**Explanation** : An RGB image contains three channels: red, green, and blue. I extract each channel using indexing, where 0 represents red, 1 represents green, and 2 represents blue. Then I display each channel in grayscale to see where each color is strongest in the image.

**Brightness Analysis** :

```python
import numpy as np                        # Import NumPy
average = np.mean(img_array)              # Calculate average brightness of the whole image
red = np.mean(img_array[:, :, 0])         # Calculate average brightness of Red channel
green = np.mean(img_array[:, :, 1])       # Calculate average brightness of Green channel
blue = np.mean(img_array[:, :, 2])        # Calculate average brightness of Blue channel
print("Average brightness:", average)     # Display overall average brightness
print("Red brightness:", red)             # Display Red channel average
print("Green brightness:", green)         # Display Green channel average
print("Blue brightness:", blue)           # Display Blue channel average
```

**Explanation** : First, I use np.mean() to find the average brightness of the whole image. Then I separately select the red, green, and blue channels and calculate their average values. The channel with the highest average value is the brightest channel

```python
import numpy as np                         # Import NumPy
import matplotlib.pyplot as plt             # Import Matplotlib
gray = np.mean(img_array, axis=2)           # Convert RGB image to grayscale by averaging R, G, B
binary = np.where(gray > 128, 255, 0)       # If brightness is above 128, make it white; otherwise black
plt.imshow(binary, cmap="gray")             # Display the black-and-white image
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the result
```

**Explanation**: I convert the RGB image into grayscale by taking the average of the three color channels. Then I use a threshold of 128. Pixels above 128 become white, and pixels below or equal to 128 become black. Finally, I display the binary image
