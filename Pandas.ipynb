#Pandas

**Creating an empty dataframe :**

```python
import pandas as pd
df=pd.DataFrame()
print(df)
```

**Creating dataframe using list :**

```python
import pandas as pd
x=['Vijay','Sandeep','Rakesh']
df=pd.DataFrame(x)
df
```

**Creating dataframe from Dict of ndarray/Lists**

```python
import pandas as pd
x={'ID':[1001,1002,1003],'Name':['Vijay','Sandeep','Rakesh']}
df=pd.DataFrame(x)
df
```

**Creating dataframe from Dict of series**

```python
import pandas as pd
x={'one':pd.Series([1,2,3,4,5],index=['a','b','c','d','e']),'two':pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'])}
df=pd.DataFrame(x)
df
```

**Column Selection**

```python
import pandas as pd
x={'one':pd.Series([1,2,3,4,5],index=['a','b','c','d','e']),'two':pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'])}
df=pd.DataFrame(x)
print(df['one'])
print(df['two'])
```

**Column Addition**

```python
import pandas as pd
x={'one':pd.Series([1,2,3,4,5],index=['a','b','c','d','e']),'two':pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'])}
df=pd.DataFrame(x)
print(df)
print("Add new column")
df['three']=pd.Series([10,20,30,40,50,60,70],index=['a','b','c','d','e','f','g'])
print(df)
print("Add new column using existing column")
df['four']=df['one']+df['three']
print(df)
```

**Column Deletion**

```python
import pandas as pd
x={'one':pd.Series([1,2,3,4,5],index=['a','b','c','d','e']),'two':pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'])}
df=pd.DataFrame(x)
print(df)
print('delete column')
del df['one']
print(df)
print('pop column')
df.pop('two')
print(df)
```

**Selection by Label**

```python
import pandas as pd
x={'one':pd.Series([1,2,3,4,5],index=['a','b','c','d','e']),'two':pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'])}
df=pd.DataFrame(x)
print(df)
print("a elements are")
print(df.loc['a'])
print("b elements are")
print(df.loc['b'])
```
